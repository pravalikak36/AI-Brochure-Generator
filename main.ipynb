```python
def fetch_website_content(url):
    response = requests.get(url)

    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    text = soup.get_text(separator=" ", strip=True)

    return text
```

```python
content = fetch_website_content("https://huggingface.co")

print(content[:2000])
```

```python
from urllib.parse import urljoin

def fetch_website_links(url):
    response = requests.get(url)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    links = soup.find_all("a")

    return [
        urljoin(url, link.get("href"))
        for link in links
        if link.get("href")
    ]
```

```python
links = fetch_website_links("https://huggingface.co")

print(len(links))
```

```python
link_system_prompt = """
You are given a list of links from a company's website.

Your task is to identify the links that are relevant for understanding
the company and creating a company brochure.

Return ONLY valid JSON in this format:

{
    "links": [
        {
            "type": "about page",
            "url": "https://example.com/about"
        }
    ]
}

Only include links that provide useful information about the company,
such as about, products, services, solutions, company information,
or careers.

Do not include privacy policy, terms of service, login, signup,
or unrelated external links.
"""
```

```python
def get_links_user_prompt(url):
    links = fetch_website_links(url)

    user_prompt = f"""
Here are the links found on {url}.

Identify which links are relevant for understanding the company
and creating a company brochure.

Links:
{chr(10).join(links)}
"""
    
    return user_prompt
```

```python
print(get_links_user_prompt("https://huggingface.co"))
```

```python
MODEL="gemini-3.5-flash"
```

```python
response = client.models.generate_content(
    model=MODEL,
    contents=link_system_prompt + "\n\n" + get_links_user_prompt(
        "https://huggingface.co"
    ),
    config={
        "response_mime_type": "application/json"
    }
)

print(response.text)
```

```python
import json

result = json.loads(response.text)

print(type(result))
print(result["links"][:3])
```

```python
selected_url = result["links"][0]["url"]

content = fetch_website_content(selected_url)

print(content[:2000])
```

```python
pages = []

for link in result["links"]:
    url = link["url"]
    content = fetch_website_content(url)

    pages.append({
        "type": link["type"],
        "url": url,
        "content": content
    })
```

```python
print(len(pages))
```

```python
print(pages[0]["type"])
print(pages[0]["url"])
print(pages[0]["content"][:350])
```

```python
def build_company_context(pages):
    context = ""

    for page in pages:
        context += f"""
PAGE TYPE: {page["type"]}
URL: {page["url"]}

CONTENT:
{page["content"]}

--------------------
"""

    return context
```

```python
company_context = build_company_context(pages)

print(company_context[:3000])
```

```python
brochure_system_prompt = """
You are an expert marketing copywriter.

Your task is to create a professional sales brochure for a company
using the information provided from its website.

The brochure should be useful for potential customers, investors,
or business partners.

Include:
- Company overview
- Products or services
- Key capabilities
- What makes the company valuable
- Relevant company information

Do not invent facts.
Only use information supported by the provided website content.

Write the brochure in clear, professional Markdown.
"""
```

```python
brochure_user_prompt = f"""
Create a sales brochure for the company using the website information
below.

Company website information:

{company_context}
"""
```

```python
brochure_response = client.models.generate_content(
    model=MODEL,
    contents=brochure_system_prompt + "\n\n" + brochure_user_prompt
)

brochure = brochure_response.text

print(brochure)
```
