```python
def fetch_website_content(url):
    response = requests.get(url)

    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    text = soup.get_text(separator=" ", strip=True)

    return text
```

```python
content = fetch_website_content("https://huggingface.co")

print(content[:2000])
```

```python
from urllib.parse import urljoin

def fetch_website_links(url):
    response = requests.get(url)
    response.raise_for_status()

    soup = BeautifulSoup(response.text, "html.parser")

    links = soup.find_all("a")

    return [
        urljoin(url, link.get("href"))
        for link in links
        if link.get("href")
    ]
```

```python
links = fetch_website_links("https://huggingface.co")

print(len(links))
```

```python
links = fetch_website_links("https://huggingface.co")

print(links[:20])
```

```python
link_system_prompt = """
You are given a list of links from a company's website.

Your task is to identify the links that are relevant for understanding
the company and creating a company brochure.

Return ONLY valid JSON in this format:

{
    "links": [
        {
            "type": "about page",
            "url": "https://example.com/about"
        }
    ]
}

Only include links that provide useful information about the company,
such as about, products, services, solutions, company information,
or careers.

Do not include privacy policy, terms of service, login, signup,
or unrelated external links.
"""
```

```python
def get_links_user_prompt(url):
    links = fetch_website_links(url)

    user_prompt = f"""
Here are the links found on {url}.

Identify which links are relevant for understanding the company
and creating a company brochure.

Links:
{chr(10).join(links)}
"""
    
    return user_prompt
```

```python
print(get_links_user_prompt("https://huggingface.co"))
```

```python
MODEL="gemini-3.5-flash"
```

```python
response = client.models.generate_content(
    model=MODEL,
    contents=link_system_prompt + "\n\n" + get_links_user_prompt(
        "https://huggingface.co"
    ),
    config={
        "response_mime_type": "application/json"
    }
)

print(response.text)
```
